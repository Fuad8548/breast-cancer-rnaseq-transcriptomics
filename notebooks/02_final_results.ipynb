# Project 1 — Final Results

## Abstract

This study examined paired T1 and T3 transcriptomic profiles from 24 patients
using a paired DESeq2 framework.

At the individual-gene level, no transcript reached the predefined
genome-wide FDR threshold.

However, ranked gene-set analysis revealed coordinated pathway-level changes.
T3-associated enrichment centered on cell-cycle progression, chromosome
segregation, DNA replication, DNA repair and protein synthesis, with additional
immune-associated signaling. T1-associated enrichment was characterized by
metabolic, extracellular-matrix, adhesion and tissue-interaction programs.

Representative genes were subsequently examined using patient-level paired
VST differences. KNL1, GYS1 and INSR showed relatively consistent directional
changes across patients, although these candidates remain exploratory rather
than validated biomarkers.

## Figure 1 — PCA

The PCA demonstrates substantial inter-patient heterogeneity and overlap
between T1 and T3, supporting the use of a paired statistical model.

## Figure 2 — Differential expression

No individual gene reaches the genome-wide FDR threshold. Nominally small
p-values and moderate fold changes exist, but these do not survive correction
for multiple testing.

## Figure 3 — GO GSEA

The T3-associated side is dominated by mitotic, chromosome-segregation,
replication and genome-maintenance processes. The T1-associated side contains
metabolic, extracellular and tissue-interaction programs.

## Figure 4 — KEGG GSEA

KEGG independently shows enrichment for cell cycle, DNA replication,
homologous recombination and related programs toward T3, while metabolic and
adhesion/signaling pathways are represented toward T1.

## Figure 5 — Candidate paired expression

Patient-level paired visualization shows whether candidate genes reproduce
their inferred direction across individual patients.

## Figure 6 — Candidate heatmap

The heatmap provides a compact view of the selected candidate genes across the
matched T1/T3 samples.

# Final interpretation

The dataset does not provide evidence for widespread genome-wide significant
single-gene differential expression under the paired DESeq2 model.

Instead, the principal signal is pathway-level coordination.

The most reproducible biological theme toward T3 is a proliferative and
genome-maintenance program involving cell-cycle progression, chromosome
segregation, DNA replication and DNA repair.

The T1 side is comparatively characterized by metabolic and
extracellular/tissue-interaction programs.

The candidate analyses strengthen the narrative at the patient level but
should be considered exploratory because candidate selection occurred after
inspection of the same dataset.

## Main limitation

Pathway enrichment from a single cohort provides hypothesis-generating
evidence. Independent validation in another cohort or experimental system
would be required before treating the candidate genes or pathway changes as
validated biomarkers or mechanistic conclusions.
